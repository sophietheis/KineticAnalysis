# Import 

```python
import os

import numpy as np
import pandas as pd

import matplotlib.pyplot as plt

from kineticanalysis.analysis.combination import combine_tracks_df_ensemble
```

# Input value

```python
path = ""
f_name = "data.csv"

# nb of combine track together
nb_track = 5

# nb of track created
nb_combination = 15
```

# Combine tracks

```python
df = pd.read_csv(os.path.join(path, f_name))
```

```python
results = combine_tracks_df_ensemble(df, nb_combination, nb_track)
```

```python
results.to_csv(os.path.join(path, "combine_track.csv"))
```

```python
fig, ax = plt.subplots()
for i in results:
    ax.plot(i["tau"], i["G"])
    
```
